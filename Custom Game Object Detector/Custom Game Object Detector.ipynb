#    Training

```python
from ultralytics import YOLO

# 1. Load a pre-trained base model
model = YOLO("yolov8n.pt") 

# 2. Train the model on our custom game dataset
if __name__ == '__main__':
    print("Starting the training process...")
    results = model.train(
        data="data.yaml",      # The configuration file we created in Level 1
        epochs=50,             # Number of times the model will analyze the whole dataset
        imgsz=640,             # Resize images to 640x640 for consistency
        batch=8,               # Process 8 images at a time (adjust based on GPU memory)
        device="cuda",         # Use the RTX 4050 GPU for maximum speed
        plots=True             # Generate performance graphs 
    )
    print("Training completed! The custom model is saved in: runs/detect/train/weights/best.pt")
```

#    Testing & Saving the weights

```python
import cv2
from ultralytics import YOLO

# 1. Load your CUSTOM trained model
# 'best.pt' contains the weights that got the highest accuracy during training
print("Loading custom game model...")
model = YOLO("runs/detect/train/weights/best.pt")

# 2. Load a gameplay video (or replace with 0 for webcam)
video_path = "resident_evil_gameplay.mp4" 
cap = cv2.VideoCapture(video_path)

# 3. Real-time Analysis Loop
while True:
    ret, frame = cap.read()
    if not ret:
        print("Video ended or cannot be read.")
        break

    # 4. Pass the frame to your custom model
    # conf=0.5 ensures it only draws a box if it is 50% sure
    results = model(frame, conf=0.5, verbose=False)

    # 5. Automatically draw boxes and labels on the frame
    annotated_frame = results[0].plot()

    # 6. Display the result
    cv2.imshow("Custom Game Radar", annotated_frame)

    # Exit when 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Clean up memory
cap.release()
cv2.destroyAllWindows()
```
